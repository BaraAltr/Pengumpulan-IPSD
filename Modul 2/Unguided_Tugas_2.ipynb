```python
import pandas as pd
import numpy as np
```

```python
movie = pd.read_csv('/content/movie_classification.csv')
movie.head(10)
```

```python
movie.mean = movie.copy(deep=True) # jika imputasi dengan mean
movie.modus = movie.copy(deep=True) # jika imputasi dengan modus
movie.median = movie.copy(deep=True) # jika imputasi dengan median
```

```python
# Mengisi missing value di kolom Time_taken dengan mean
movie.mean.fillna({'Time_taken': movie.mean['Time_taken'].mean()}, inplace=True)
# Mengisi missing value di kolom Time_taken dengan modus
movie.modus.fillna({'Time_taken': movie.modus['Time_taken'].mode()[0]}, inplace=True)
# Mengisi missing value di kolom Time_taken dengan median
movie.median.fillna({'Time_taken': movie.median['Time_taken'].median()}, inplace=True)
```

```python
movie.mean.isnull().sum()
```

```python
movie.modus.isnull().sum()
```

```python
movie.median.isnull().sum()
```

```python
# Simpan data ke csv
movie.mean.to_csv('/content/movie_classification.csv', index=False)
movie.modus.to_csv('/content/movie_classification.csv', index=False)
movie.median.to_csv('/content/movie_classification.csv', index=False)
```

```python
# Simpan data ke exel
movie.mean.to_excel('/content/movie_classification.xlsx', index=False)
movie.modus.to_excel('/content/movie_classification.xlsx', index=False)
movie.median.to_excel('/content/movie_classification.xlsx', index=False)
```
